```python
import requests
import pandas as pd

latitude = 18.5204
longitude = 73.8567

url = f"https://archive-api.open-meteo.com/v1/archive?latitude={latitude}&longitude={longitude}&start_date=2023-01-01&end_date=2024-12-31&daily=temperature_2m_max,temperature_2m_min,precipitation_sum,wind_speed_10m_max&timezone=auto"

response = requests.get(url)

data = response.json()

df = pd.DataFrame({
    "Date": data["daily"]["time"],
    "Temp_Max": data["daily"]["temperature_2m_max"],
    "Temp_Min": data["daily"]["temperature_2m_min"],
    "Rainfall": data["daily"]["precipitation_sum"],
    "WindSpeed": data["daily"]["wind_speed_10m_max"]
})

df.head()
```

```python
# Train/Test Split + Validation
```

```python
from prophet import Prophet
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
import numpy as np

# Prepare data for Prophet

prophet_df = df[["Date","Temp_Max"]].copy()

prophet_df.columns = ["ds","y"]

prophet_df["ds"] = pd.to_datetime(prophet_df["ds"])

# 80% Train 20% Test Split

split_index = int(len(prophet_df)*0.8)

train_df = prophet_df.iloc[:split_index]

test_df = prophet_df.iloc[split_index:]

print("Training Records:", len(train_df))
print("Testing Records:", len(test_df))
```

```python
# Why Train-Test Split?

# ans=To evaluate forecasting performance fairly, the dataset was divided into training and testing subsets. The model learns patterns from historical climate data using the training set and is then evaluated on unseen test data.

# Training Data= 80% of the historical climate observations were used to train the Prophet forecasting model.

# Testing Data= The remaining 20% of the dataset was reserved for validation and performance evaluation.

# Validation Goal= This approach helps assess how accurately the model can predict future temperatures on data that was not used during training.
```

```python
# Train Prophet Model
```

```python
model = Prophet()

model.fit(train_df)
```

```python
# Predict On Test Period
```

```python
future = model.make_future_dataframe(
    periods=len(test_df)
)

forecast = model.predict(future)
```

```python
# Validation
```

```python
predictions = forecast.tail(len(test_df))

actual = test_df["y"].values

predicted = predictions["yhat"].values
```

```python
# Metrics
```

```python
mae = mean_absolute_error(
    actual,
    predicted
)

rmse = np.sqrt(
    mean_squared_error(
        actual,
        predicted
    )
)

mape = np.mean(
    np.abs(
        (actual-predicted)/actual
    )
)*100

r2 = r2_score(
    actual,
    predicted
)

print("MAE :", round(mae,2))
print("RMSE :", round(rmse,2))
print("MAPE :", round(mape,2))
print("R2 Score :", round(r2,2))
```

```python
# Climate Risk Detection
```

```python
def climate_risk(temp, rainfall):

    if temp >= 45:
        return "Severe Heatwave"

    elif temp >= 40:
        return "Moderate Heatwave"

    elif rainfall >= 100:
        return "Extreme Rainfall"

    elif rainfall >= 50:
        return "Heavy Rainfall"

    else:
        return "Normal"


df["Risk_Level"] = df.apply(
    lambda row:
    climate_risk(
        row["Temp_Max"],
        row["Rainfall"]
    ),
    axis=1
)

df[["Temp_Max","Rainfall","Risk_Level"]].head()
df.head()
```

```python
# Risk detection was enhanced by introducing multiple climate risk categories.
# The system now classifies climate conditions into Moderate Heatwave, Severe Heatwave, Heavy Rainfall, and Extreme Rainfall based on temperature and rainfall intensity.
```

```python
# Why Used?
# To detect abnormal climate conditions such as heatwaves and extreme rainfall events.

# What Does It Do?
# It identifies weather conditions that exceed predefined risk thresholds.

# Conclusion
# Risk detection improves environmental awareness and enables early warning generation.
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

actual = forecast_df["y"][-100:]

predicted = forecast["yhat"][:len(actual)]

mae = mean_absolute_error(actual,predicted)

rmse = mean_squared_error(actual,predicted) ** 0.5

r2 = r2_score(actual,predicted)

print("MAE:",mae)
print("RMSE:",rmse)
print("R2:",r2)
```

```python
# Metric 1 — MAE (Mean Absolute Error)
# Meaning:
# MAE measures the average difference between actual and predicted values.

# Interpretation= A lower MAE indicates better prediction accuracy.
```

```python
# Metric 2 — RMSE (Root Mean Squared Error)
# Meaning: RMSE measures prediction errors while giving higher importance to larger errors.

# Interpretation:A lower RMSE value indicates that the model's predictions are closer to actual observations.
```

```python
# Metric 3 — R² Score
# Meaning: R² Score measures how much variation in the data is explained by the model.

# Interpretation: Values closer to 1 indicate better model performance and stronger predictive capability.

# Conclusion: A higher R² Score suggests that the forecasting model effectively captures climate patterns and trends.
```
